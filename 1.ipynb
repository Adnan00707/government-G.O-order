```python
import os
import re
import pandas as pd

# Define path to test1 folder
test1_dir = './test1'

# Regex pattern for Kerala Government Order (G.O.) Numbers
go_pattern = r'G\.?\s*O\.?\s*(?:\(?\s*(?:Rt|Ms|P)\.?\s*\)?\s*)?(?:No\.|No:|No)\s*[:.]?\s*[\d/A-Za-z.-]+'

results = []

# Iterate through all files inside the test1 folder
files = sorted([f for f in os.listdir(test1_dir) if f.endswith('.md')])

for filename in files:
    filepath = os.path.join(test1_dir, filename)
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
        
    lines = content.split('\n')
    extracted_go = None
    
    # Process the top 30 lines (header block) to extract the primary G.O. Number
    for line in lines[:30]:
        # Clean common markdown signs
        cleaned_line = line.replace('#', '').replace('*', '').replace('_', '').strip()
        
        # Skip reference sections to avoid extracting reference order numbers
        if any(x in cleaned_line.upper() for x in ['READ', 'LETTER', 'U.O. NOTE', 'WP(C)', 'JUDGEMENT', 'REFERENCE']):
            continue
            
        m = re.search(go_pattern, cleaned_line, re.IGNORECASE)
        if m:
            extracted_go = m.group(0).strip()
            break
            
    results.append({
        'Filename': filename,
        'Extracted GO Number': extracted_go
    })

# Convert to DataFrame
df = pd.DataFrame(results)

# Display the DataFrame
df
```
